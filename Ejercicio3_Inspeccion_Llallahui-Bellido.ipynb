## Ejercicio 3: Inspección visual de un producto

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Matriz NumPy de 5x5 píxeles, donde 0 representa un píxel normal y 1 representa un píxel defectuoso. |
| **Acciones (A)** | Aprobar el producto, enviarlo a revisión manual o rechazarlo. |
| **Entorno (E)** | Línea de producción, productos, cámara de inspección y área de control de calidad. |
| **Objetivo** | Detectar productos con defectos y evitar que productos con fallas importantes continúen en el proceso de producción. |
| **Medida de desempeño (P)** | Detectar correctamente los defectos, reducir productos defectuosos aprobados y evitar revisiones manuales innecesarias. |

### Justificación de las reglas

Se aplicarán las siguientes reglas de decisión:

- Si se detectan 0 píxeles defectuosos, el producto se aprueba.
- Si se detectan entre 1 y 3 píxeles defectuosos, el producto se envía a revisión manual.
- Si se detectan más de 3 píxeles defectuosos, el producto se rechaza.

Estas reglas permiten diferenciar los productos sin fallas de aquellos que presentan defectos leves o graves. Los casos con pocos defectos se envían a revisión manual para evitar rechazos innecesarios, mientras que una cantidad mayor de defectos indica un problema de calidad suficiente para rechazar el producto.

### Código

```python
import numpy as np

def agente_inspeccion(imagen):
    """
    Percepción (S): matriz 5x5 de NumPy
    0 = píxel normal
    1 = píxel defectuoso
    """

    defectos = np.sum(imagen == 1)

    if defectos == 0:
        return "aprobar", "no se detectaron píxeles defectuosos"

    elif defectos <= 3:
        return "revisión manual", f"se detectaron {defectos} píxeles defectuosos"

    else:
        return "rechazar", f"se detectaron {defectos} píxeles defectuosos"
```
