## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Tiempo de espera en minutos, nivel de urgencia del ticket y condición de cliente premium. |
| **Acciones (A)** | Mantener el ticket en nivel 1, escalar a nivel 2 o escalar a nivel 3. |
| **Entorno (E)** | Mesa de ayuda, clientes, tickets de soporte, técnicos y sistema de atención. |
| **Objetivo** | Atender los tickets de acuerdo con su prioridad y evitar tiempos de espera excesivos. |
| **Medida de desempeño (P)** | Reducir el tiempo de atención, priorizar correctamente los casos urgentes y mejorar la satisfacción del cliente. |

### Justificación de las reglas

Se aplicarán las siguientes reglas de decisión:

- Si la urgencia es alta, el ticket se escala directamente a nivel 3.
- Si la urgencia es media y el tiempo de espera es mayor o igual a 30 minutos, se escala a nivel 2.
- Si la urgencia es media y el cliente es premium, también se escala a nivel 2.
- Si la urgencia es baja y el tiempo de espera es mayor o igual a 60 minutos, se escala a nivel 2.
- Si la urgencia es baja, pero el cliente es premium y espera 30 minutos o más, se escala a nivel 2.
- En los demás casos, el ticket se mantiene en nivel 1.

Estas reglas permiten atender primero los casos más críticos y evitar tiempos de espera excesivos. También se considera la prioridad de los clientes premium sin descuidar la urgencia real del problema. De esta manera, el equipo de soporte puede distribuir mejor su carga de trabajo y ofrecer una atención más rápida y ordenada.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):

    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta, requiere atención inmediata"

    elif nivel_urgencia == "media":
        if tiempo_espera_minutos >= 30 or cliente_premium:
            return "escalar a nivel 2", "urgencia media con tiempo de espera elevado o cliente premium"
        else:
            return "mantener en nivel 1", "urgencia media con tiempo de espera menor a 30 minutos"

    elif nivel_urgencia == "baja":
        if tiempo_espera_minutos >= 60:
            return "escalar a nivel 2", "urgencia baja, pero el tiempo de espera es elevado"
        elif cliente_premium and tiempo_espera_minutos >= 30:
            return "escalar a nivel 2", "cliente premium con tiempo de espera prolongado"
        else:
            return "mantener en nivel 1", "urgencia baja y tiempo de espera dentro del rango normal"
```

### Simulación y pruebas

```python
casos_prueba = [
    (5, "alta", False),
    (10, "media", False),
    (30, "media", False),     # caso límite de tiempo
    (15, "media", True),
    (20, "baja", False),
    (35, "baja", True),       # urgencia baja, premium y espera alta
    (70, "baja", False)       # urgencia baja, pero espera excesiva
]

for tiempo, urgencia, premium in casos_prueba:
    accion, motivo = agente_soporte(tiempo, urgencia, premium)

    print("Tiempo de espera:", tiempo, "minutos")
    print("Nivel de urgencia:", urgencia)
    print("Cliente premium:", premium)
    print("Decisión:", accion)
    print("Motivo:", motivo)
    print("-" * 50)
```
